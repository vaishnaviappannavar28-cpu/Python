Merge, Joins, and Concatenate

```python
import pandas as pd
```

```python
df1 = pd.read_excel(r"C:/Users/hp/Downloads/customers.xlsx")
df1
```

```python
df2 = pd.read_excel(r"C:/Users/hp/Downloads/orders.xlsx")
df2
```

```python
df1.merge(df2)
```

```python
df1.merge(df2 , how = 'inner' , on = 'Customer_ID')
```

```python
df1.merge(df2 , how = 'outer')
```

```python
df1.merge(df2 , how = 'left')
```

```python
df1.merge(df2 , how = 'right')
```

```python
df1.merge(df2 , how = 'cross')
```

```python
df1.join(df2 , on = 'Customer_ID', how = 'outer', lsuffix= '_Left', rsuffix='_Right')
```

```python
df3 = df1.set_index('Customer_ID').join(df2.set_index('Customer_ID'),lsuffix='_Left',rsuffix='_Right', how='inner')
df3
```

```python
pd.concat([df1,df2], join='outer',axis=1)
```
